```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mirichoi0218/insurance")

print("Path to dataset files:", path)
```

```python
import os

# Listar arquivos dentro da pasta
print(os.listdir(path))
```

```python
import pandas as pd

file_path = os.path.join(path, 'insurance.csv')
df = pd.read_csv(file_path)
print(df.head())
```

```python
df.sample(10)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Melhor visualização dos gráficos
sns.set(style="whitegrid")
```

```python
df.shape         # Linhas e colunas
```

```python
df.columns       # Nomes das colunas
```

```python
df.info()        # Tipos de dados e valores nulos
```

```python
df.describe()    # Estatísticas para colunas numéricas
```

```python
df.isnull().sum()         # Ver quantos valores nulos por coluna
```

```python
df.duplicated().sum()     # Ver se há linhas duplicadas
```

```python
df['charges'].hist(bins=30)
plt.title("Distribuição dos custos de seguro")
plt.xlabel("Valor ($)")
plt.ylabel("Número de pessoas")
plt.show()
```

```python
sns.boxplot(y=df['bmi'])
plt.title("Boxplot do IMC")
plt.show()
```

```python
df['smoker'].value_counts().plot(kind='bar')
plt.title("Fumantes vs Não fumantes")
plt.ylabel("Quantidade")
plt.show()
```

```python
sns.boxplot(x='smoker', y='charges', data=df)
plt.title("Custo de seguro por hábito de fumar")
plt.show()
```

```python
sns.scatterplot(x='bmi', y='charges', data=df)
plt.title("IMC vs Custo de seguro")
plt.show()
```

```python
sns.pairplot(df, hue='smoker')
plt.show()

# Correlação entre variáveis numéricas
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Mapa de calor das correlações")
plt.show()
```

```python
sns.boxplot(x='charges', data=df)
plt.title("Boxplot dos custos médicos")
plt.show()
```
